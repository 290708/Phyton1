Orden Ascendente

```python
import sys
numbers = [5, 2, 9, 1, 7]
numbers.sort()
sys.stdout.write("11. " + str(numbers) + "\n")
```

Orden Decendente

```python
import sys
numbers = [5, 2, 9, 1, 7]
numbers.sort(reverse=True)
sys.stdout.write("12. " + str(numbers) + "\n")
```

Multiplicar Con el numero 2

```python
import sys
numbers = [3, 6, 9, 12, 15]
new_list = [x * 2 for x in numbers]
sys.stdout.write("13. " + str(new_list) + "\n")
```

Imprimir Cada Palabra en Linea Separada

```python
import sys
words = ["python", "list", "exercise", "practice"]
sys.stdout.write("")
for word in words:
    sys.stdout.write(word + "\n")
```

Calcula el Promedio

```python
import sys
grades = [80, 75, 90, 85, 95]
average = sum(grades) / len(grades)
sys.stdout.write("15. " + str(average) + "\n")
```
